## III. Méthode de gradient stochastique (SGD)

La descente de gradient stochastique (SGD) apporte une nuance à l'approche classique de la descente de gradient. Le terme "stochastique" se réfère à un système ou à un processus lié à une probabilité aléatoire. Par conséquent, ce caractère aléatoire est intégré dans le calcul du gradient, ce qui altère significativement son comportement et son efficacité par rapport à la descente de gradient standard.

Dans la descente de gradient traditionnelle par lots, le gradient de la fonction de perte est calculé par rapport aux paramètres de l'ensemble de l'entraînement. Comme on peut l'imaginer, pour de grands ensembles de données, cela peut être assez coûteux en termes de calcul et de temps. C'est là que SGD entre en jeu. Au lieu d'utiliser l'intégralité de l'ensemble de données pour calculer le gradient, SGD sélectionne de manière aléatoire un seul point de données (ou quelques points de données) pour calculer le gradient à chaque itération.

1. Choisissez un point de départ (initialisation).
2. **Calculez le gradient à cet emplacement en sélectionnant au hasard un seul point de données (ou un petit lot de points de données) dans l'ensemble de données.**
3. Effectuez un pas proportionnel dans la direction opposée (négative) du gradient.
4. Répétez les étapes 2 et 3 jusqu'à ce que le gradient mis à l'échelle soit très petit ou qu'un nombre maximum d'itérations soit atteint.

## IV. Comparaison entre BGD et SGD

Nous comparons maintenant la vitesse de convergence des algorithmes BGD et SGD dans la situation actuelle en comparant les temps de calcul.

En Python, nous utilisons **time.time()** pour lire l'heure du système.

```python
from time import time
tics = []
def tic():
    tics.append(time())
    
def toc():
    if len(tics) == 0:
        return None
    else:
        return time()-tics.pop()
```

```python
tic()
params = dict(fonc_gradient = gradient, taux_apprenti = 0.1, nom_iter = 100)
X_liste = BGD(X_org = (10,10), **params)
toc()
```

### PyTorch

```python
import torch

x = torch.tensor([1.0, 2.0], requires_grad=True)
optim = torch.optim.SGD([x], lr=taux_apprend)

for it in range(nom_iter):
    optim.zero_grad()
    y = g(x)
    y.backward()
    optim.step()
    if (it+1) % 100 == 0:
        print(f'itération [{it+1}/{nom_iter}], g(x,y): {y.item()}')
print("Minimum:", x.detach().numpy())
```

### TensorFlow

```python
import tensorflow as tf

def target_function(x1, x2):
    return x1**2 + x2**2 + x1*x2

x1 = tf.Variable(1.0, dtype=tf.float32)
x2 = tf.Variable(2.0, dtype=tf.float32)

optimizer = tf.optimizers.SGD(taux_apprend)

# 迭代寻找最小值
for it in range(nom_iter):
    with tf.GradientTape() as tape:
        y = target_function(x1, x2)
    gradients = tape.gradient(y, [x1, x2])
    optimizer.apply_gradients(zip(gradients, [x1, x2]))
    if (it+1) % 100 == 0:
        print(f'itération [{it+1}/{nom_iter}], g(x,y): {target_function(x1, x2)}')
print("Minimum:", [x1.numpy(), x2.numpy()])
```
