#### 4. Momentum

Lorsque la vallée est très étroite : la descente de gradient oscille.

On ajoute donc une inertie :
$$v_{k+1}=\beta. v_k-\eta. \nabla J(x_k)$$

$$x_{k+1}=x_k+v_{k+1}$$

Pour une fonction quadratique :
$$f(x)=\frac{1}{2} x^\top \cdot A \cdot x$$

la vitesse asymptotique est (avec $\kappa=\frac{\lambda_{max}}{\lambda_{min}}$):
- Gradient :
    $$\rho_{GD}=\frac{\kappa-1}{\kappa+1}$$
- Momentum :
    $$\rho_{Mom}=\frac{\sqrt{\kappa}-1}{\sqrt{\kappa}+1}$$

#### 3.3. Méthode de Newton

La méthode de Newton utilise :
- le gradient
- la matrice Hessienne
afin d'obtenir une approximation quadratique locale.

Autour de $x_k$ :
$$f(x) \approx f(x_k)+\nabla f(x_k)^\top\cdot(x-x_k)+\frac{1}{2}(x-x_k)^\top \cdot H(x_k)\cdot (x-x_k)$$

En annulant le gradient :
$$\nabla f(x_k)+H(x_k)\cdot p_k=0$$

on obtient :
$$p_k=-H(x_k)^{-1}\cdot\nabla f(x_k)$$

Si
$$f(x)=\frac{1}{2}x^\top\cdot A\cdot x - b^\top \cdot x$$
alors :
$$H(x)=A$$

Newton converge en : UNE SEULE itération.

**Vitesse de convergence** : $\|e_{k+1}\| \leq C\|e_k\|^2$

##### 3.4.2.2 Formule BFGS

BFGS est aujourd'hui la méthode quasi-Newton la plus utilisée.

**Mise à jour BFGS** :
$$H_{k+1}=(I-\rho s_k\cdot y_k^T)\cdot H_k \cdot (I-\rho y_k\cdot s_k^T)+\rho s_k\cdot s_k^T$$
avec :
$$\rho=\frac{1}{y_k^T\cdot s_k}$$


BFGS approxime :

##### 3.4.3 Approfondissement

Dans de nombreuses applications modernes telles que l'apprentissage automatique, l'optimisation de trajectoires de drones, la vision par ordinateur ou les systèmes de recommandation, le nombre de variables à optimiser peut atteindre plusieurs dizaines de milliers, voire plusieurs millions.

Nous avons étudié plusieurs méthodes d'optimisation :
- la descente de gradient ;
- la méthode de Newton ;
- la méthode quasi-Newton BFGS.

La méthode de Newton nécessite le calcul et l'inversion de la matrice hessienne, ce qui devient rapidement prohibitif lorsque la dimension du problème augmente.

La méthode BFGS évite le calcul explicite de la hessienne en construisant une approximation de son inverse :
$$H_k \approx \left(H_f(x_k)\right)^{-1}$$
Cependant, cette méthode doit toujours stocker une matrice dense de taille $n \times n$.

Lorsque
$$n = 10^4,\quad 10^5,\quad 10^6, \quad \ldots$$
la mémoire nécessaire devient considérable.
