## Sinusoidal generator (Euler explicite)

```python
from numpy import *
import matplotlib.pyplot as plt
```

```python
k=.05
N=1000

x=zeros(N) 
y=zeros(N)
t=range(N);

x[0]=1
```

```python

for i in range(N-1):
  x[i+1] = x[i] + k * y[i];
  y[i+1] = y[i] - k * x[i];
```

Figures
 

```python
plt.plot(t,x,'r',y,'b')
plt.title("x et y en fonction du temps")
plt.xlabel('t')
plt.ylabel('x et y')
plt.legend(['x','y'])
plt.grid()


```

```python
plt.plot(x,y,'k')
plt.title("Plan de phase")
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
```
