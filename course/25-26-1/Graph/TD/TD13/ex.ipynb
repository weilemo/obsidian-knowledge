```python
import numpy as np
```

```python
inf = float ( ' inf ')
```

```python
def initial (n, s):
    Satraiter = [i for i in range(n) if i != s]

    Poids = [np.inf] * n
    Poids[s] = 0
    
    Parent = [np.inf] * n
    Parent[s] = 'non'

    return Satraiter, Poids, Parent
```

```python
def voisin(M, k, L):
    u = []
    n = len(M)  
    for i in L:
        if M[k, i] != np.inf and i != k:
            u.append(i)
    return u
```

```python
def echange(M, k, j, po, pa):
    """
    松弛操作
    M: 邻接矩阵
    k: 中转点
    j: 目标点
    po: 权重列表 (Poids)
    pa: 父节点列表 (Parent)
    """
    # 按照题目描述的逻辑判断
    # M[k, j] 就是题目中的 v(k, j)
    if po[j] > po[k] + M[k, j]:
        po[j] = po[k] + M[k, j]  # 更新距离
        pa[j] = k                # 更新前驱(父节点)
    
    # 题目要求返回更新后的列表
    return po, pa
```

```python
def Dijkstra(M, s):
    n = len(M)
    # 调用之前的 initial 函数
    Satraiter, Poids, Parent = initial(n, s)
    k = s
    
    # 主循环：只要待处理列表不为空
    while Satraiter != []:
        # 【空1】找出 k 在 Satraiter 中的邻居
        V = voisin(M, k, Satraiter)
        
        # 对每个邻居进行松弛操作
        for j in V:
            echange(M, k, j, Poids, Parent)
            
        # --- 下面这一段是在 Satraiter 中找 Poids 最小的点 ---
        r = 0
        j = Satraiter[r]
        smin = j            # 假设第一个就是最小的
        m = Poids[smin]     # 记录最小权重
        
        # 【空2】遍历 Satraiter 列表
        while r < len(Satraiter) - 1:
            r = r + 1
            j = Satraiter[r]
            # 【空3】如果发现比当前 m 更小的距离
            if Poids[j] < m:
                smin = j
                m = Poids[j]
        
        # 将找到的最近的点移出待处理列表
        Satraiter.remove(smin)
        
        # 【空4】将当前点更新为 smin，作为下一轮的起点
        k = smin
        
    return Poids, Parent
```
