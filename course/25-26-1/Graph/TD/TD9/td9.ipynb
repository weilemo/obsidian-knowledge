```python
import matplotlib . pyplot as plt
import networkx as nx
import numpy as np
```

```python

D = {'0':['1','2','4'],'1':['0','2','3'],'2':['0','1','3','4','6'],'3':['1','2','5'],'4':['0','2'],'5':['3','6'],'6':['2','5']}
G = nx . Graph ( D ) # ou nx . DiGraph ( D )
s = '5'
Gdfs = nx . dfs_tree (G , source = s )
nx . draw ( Gdfs , with_labels = True )
plt . show ()
```

```python
def ParcoursDFS (G , s ) :
    Access =[ s ]
    Avisiter =[ s ]
    while Avisiter !=[]:
        u = Avisiter . pop ()
        L = list ( G [ u ])
        d = len ( L )
        k = 0
        while k < d :
            x = L [ k ]
            if not ( x in Access ) :
                Avisiter . append (u)
                Avisiter . append (x)
                Access . append (x)
                k = d
            k = k + 1
    return ( Access )

ParcoursDFS(G,'5')
```

```python
def parcoursDFS2(M, s):
    n = len(M)  
    Access = [s]
    Avisiter = [s]
    while Avisiter !=[]:
        u = Avisiter.pop()
        for v in range(n):
            if M[u][v] != 0 and v not in Access:
                Avisiter.append(u)    
                Avisiter.append(v)    
                Access.append(v)      
                break                 
    return Access
parcoursDFS2(M,'5')
```
