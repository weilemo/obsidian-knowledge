```python
import numpy as np
def Mat(n):
    matrix = np.zeros((n,n))
    for i in range(n):
        matrix[i,0:i+1]=1/(i+1)
    return matrix

Mat(5)
```

```python
def Mat2(n):
    matrix = np.zeros((n,n))
    for i in range(n-1):
        matrix[i,i+1]=1
        matrix[i+1,i]=1
    return matrix

Mat2(5)
```

```python
def matM(x):
    n=len(x)
    matrix = np.zeros((n,2))
    for i in range(n):
        matrix[i,0]=x[i]
        matrix[i,1]=0
    return matrix
matM([1,2,3,4,5])
```
